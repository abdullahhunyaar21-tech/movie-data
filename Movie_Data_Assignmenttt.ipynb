### **Question 28**
(Visualization) 'No_of_Votes' ke mutabiq top 5 highest voted movies ka ek Horizontal Bar Chart banayein.

```python
import matplotlib.pyplot as plt
import pandas as pd

print("Aapke dataset ke columns yeh hain:", list(df.columns))

possible_names = ['Title', 'Series_Title', 'Movie_Title', 'title', 'name', 'Movie_Name', 'Movie']
title_col = None
for col in possible_names:
    if col in df.columns:
        title_col = col
        break

if title_col is None:
    title_col = df.select_dtypes(include=['object']).columns[0]

print(f"\n-> Hum chart banane ke liye '{title_col}' column use kar rahe hain.\n")

df['No_of_Votes'] = pd.to_numeric(df['No_of_Votes'].astype(str).str.replace(',', ''), errors='coerce').fillna(0).astype(int)

top_voted = df.sort_values(by='No_of_Votes', ascending=False).head(5)
top_voted = top_voted.sort_values(by='No_of_Votes', ascending=True)

plt.figure(figsize=(10, 5))
plt.barh(top_voted[title_col], top_voted['No_of_Votes'], color='#6a1b9a', edgecolor='black')

plt.title('Top 5 Highest Voted Movies', fontsize=14, fontweight='bold')
plt.xlabel('Total Number of Votes', fontsize=12)
plt.ylabel('Movie Title', fontsize=12)
plt.tight_layout()
plt.show()
```

### **Question 36**
(Visualization) 'Drama' aur 'Action' genres ki movies ke 'Gross' revenue ko compare karne ke liye ek Grouped Box Plot ya Violin Plot banayein.

```python
import matplotlib.pyplot as plt
import seaborn as sns

drama_or_action = df[df['Genre'].str.contains('Drama|Action', case=False, na=False)].copy()

drama_or_action['Main_Genre'] = drama_or_action['Genre'].apply(lambda x: 'Drama' if 'Drama' in str(x) else 'Action')

drama_or_action['Gross'] = pd.to_numeric(drama_or_action['Gross'].astype(str).str.replace(',', ''), errors='coerce')

plt.figure(figsize=(8, 6))
sns.boxplot(data=drama_or_action, x='Main_Genre', y='Gross', hue='Main_Genre', palette='Set2', legend=False)

plt.title('Gross Revenue Comparison: Drama vs Action', fontsize=14, fontweight='bold')
plt.xlabel('Genre', fontsize=12)
plt.ylabel('Gross Revenue', fontsize=12)
plt.show()
```

### **Question 40**
(Visualization) Top 5 highest rated movies aur unke 'Meta_score' vs 'IMDB_Rating' (scaled to 100) ka ek Side-by-Side Comparison Bar Chart banayein.

```python
import matplotlib.pyplot as plt
import numpy as np

top_5_rated = df.sort_values(by='IMDB_Rating', ascending=False).head(5)

movie_titles = top_5_rated['Series_Title']
imdb_scaled = top_5_rated['IMDB_Rating'] * 10
meta_scores = top_5_rated['Meta_score']

x = np.arange(len(movie_titles))
width = 0.35

plt.figure(figsize=(12, 6))
plt.bar(x - width/2, imdb_scaled, width, label='IMDB Rating (Scaled to 100)', color='#ffcc00', edgecolor='black')
plt.bar(x + width/2, meta_scores, width, label='Meta Score', color='#3399ff', edgecolor='black')

plt.title('Top 5 Movies: IMDB Rating vs Meta Score Side-by-Side', fontsize=14, fontweight='bold')
plt.xlabel('Movie Title', fontsize=12)
plt.ylabel('Scores (Out of 100)', fontsize=12)
plt.xticks(x, movie_titles, rotation=25, ha='right')
plt.legend()
plt.tight_layout()
plt.show()
```
